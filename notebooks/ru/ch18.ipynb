## Стартер капстоун-проекта 3: агент с одним инструментом

Исполняемый стартер, а не распечатка шаблона: использует ограниченный AST-калькулятор главы 15 (никакого `eval`; пустые builtins НЕ изоляция) и требует от модели проверенное поле `expression`, а не сам вопрос. Демо симулирует структурированный ответ модели; для реального запуска подключите `chat()`.

```python
# Агент с одним инструментом: ОДИН инструмент, безопасно (исполняемый код).
# Использует ограниченный AST-калькулятор главы 15; модель возвращает
# проверенное поле expression, а не сам вопрос. Никакого eval().
import ast
import json
import math
import re

# ===== Chapter 15's bounded AST calculator (verbatim) =====
def _pow(a, b):
    if abs(b) > 100:  # "9**9**9" would otherwise hang the CPU for minutes
        raise ValueError("exponent too large")
    return a ** b

_OPS = {ast.Add: lambda a, b: a + b, ast.Sub: lambda a, b: a - b,
        ast.Mult: lambda a, b: a * b, ast.Div: lambda a, b: a / b,
        ast.Pow: _pow, ast.FloorDiv: lambda a, b: a // b,
        ast.Mod: lambda a, b: a % b,
        ast.UAdd: lambda a: +a, ast.USub: lambda a: -a}

def calculate(expr):
    """Safely evaluate arithmetic: no eval, no builtins, bounded work."""
    if len(expr) > 200:
        raise ValueError("expression too long")
    tree = ast.parse(expr, mode="eval")
    if sum(1 for _ in ast.walk(tree)) > 50:
        raise ValueError("expression too complex")
    def check(v):  # every intermediate must be a finite, bounded real
        if isinstance(v, bool) or not math.isfinite(v) or abs(v) > 1e12:
            raise ValueError("value out of bounds")
        return v
    def walk(node):
        if isinstance(node, ast.Expression):
            return walk(node.body)
        if isinstance(node, ast.Constant) and type(node.value) in (int, float):
            return check(node.value)  # type() excludes bool (True+True)
        if isinstance(node, (ast.BinOp, ast.UnaryOp)) and type(node.op) in _OPS:
            args = ([walk(node.left), walk(node.right)]
                    if isinstance(node, ast.BinOp) else [walk(node.operand)])
            return check(_OPS[type(node.op)](*args))
        raise ValueError("disallowed expression")
    return walk(tree)

def calculator(expression: str) -> str:
    """The ONE tool: arithmetic only, validated by the AST whitelist."""
    try:
        return str(calculate(expression))
    except (ValueError, SyntaxError, TypeError, ZeroDivisionError):
        return "Error: could not calculate"

# ===== Tool selection: ask the model for a validated expression field =====
TOOL_PROMPT = (
    "If the user asks for arithmetic, reply with ONLY a JSON object: "
    '{"tool": "calculator", "expression": "<math expression>"} '
    "where expression contains only numbers and + - * / ** % ( ). "
    'Otherwise reply {"tool": null}. Never put the question itself in expression.'
)

def parse_tool_call(model_reply: str):
    """Validate the model's structured reply; reject anything malformed."""
    try:
        call = json.loads(model_reply)
    except json.JSONDecodeError:
        return None
    if not isinstance(call, dict) or call.get("tool") != "calculator":
        return None
    expr = call.get("expression")
    if not isinstance(expr, str) or not re.fullmatch(r"[0-9+\-*/%(). eE]+", expr):
        return None  # letters other than exponent 'e' never reach the tool
    return expr

def agent(user_message: str, model_reply: str) -> str:
    """Process the model's structured decision (second pass in Ch15 style)."""
    expr = parse_tool_call(model_reply)
    if expr is None:
        return "I can help with calculations. Try: calculate 15 * 7"
    result = calculator(expr)
    return f"The answer is: {result}"

# ===== Демо: реальный LLM выдаёт model_reply; здесь мы его симулируем =====
# В продакшене: model_reply = chat(TOOL_PROMPT + "\nUser: " + question)
question = "What is 15 * 7?"
simulated_reply = '{"tool": "calculator", "expression": "15 * 7"}'
print(agent(question, simulated_reply))          # The answer is: 105

# Старый баг теперь невозможен: весь вопрос НЕ является допустимым выражением
print(agent(question, json.dumps({"tool": "calculator", "expression": question})))

# Попытки инъекции отклоняются до любого исполнения
print(calculator("__import__('os').system('ls')"))
print(parse_tool_call('{"tool": "calculator", "expression": "().__class__"}'))
```
