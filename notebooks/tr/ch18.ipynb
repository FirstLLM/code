## Bitirme Projesi 3 Başlangıç: Tek Araçlı Ajan

Ajan bitirme projesini seçtiyseniz, işte bir iskelet:

```python
# Tek Araçlı Ajan Başlangıç Şablonu
agent_template = '''
# Tek Araçlı Ajan
# Basit başlayın: YALNIZCA BİR araç!

import json

# BİR aracınızı tanımlayın
def calculator(expression: str) -> str:
    """Bir matematik ifadesini değerlendirin."""
    try:
        # UYARI: eval tehlikelidir! Üretimde uygun bir matematik ayrıştırıcısı kullanın
        result = eval(expression, {"__builtins__": {}}, {})
        return str(result)
    except:
        return "Hata: Hesaplanamadı"

TOOL_SCHEMA = {
    "name": "calculator",
    "description": "Bir matematik ifadesini değerlendir",
    "parameters": {
        "expression": "'2 + 2' veya '15 * 7' gibi bir matematik ifadesi"
    }
}

def should_use_tool(user_message: str) -> bool:
    """Hesap makinesine ihtiyacımız olup olmadığına karar verin."""
    # TODO: Sınıflandırmak için LLM kullanın (Böl15)
    math_keywords = ["hesapla", "işlem yap", "nedir", "+", "-", "*", "/"]
    return any(kw in user_message.lower() for kw in math_keywords)

def extract_expression(user_message: str) -> str:
    """Mesajdan matematik ifadesini çıkarın."""
    # TODO: Çıkarmak için LLM kullanın (Böl15)
    return user_message  # Yer tutucu

def agent(user_message: str) -> str:
    """Kullanıcı mesajını işleyin, potansiyel olarak araçları kullanın."""
    if should_use_tool(user_message):
        expression = extract_expression(user_message)
        result = calculator(expression)
        return f"Cevap: {result}"
    else:
        return "Hesaplamalarla yardımcı olabilirim. Bir şeyler hesaplamamı isteyin!"

# Test edin
print(agent("15 çarpı 7 nedir?"))
'''

print("Tek Araçlı Ajan Başlangıç Şablonu:")
print(agent_template)
```
