## Abschlussprojekt 3 Starter: Einzelwerkzeug-Agent

Wenn Sie das Agenten-Abschlussprojekt gewählt haben, hier ist ein Grundgerüst:

```python
# Einzelwerkzeug-Agent Starter-Vorlage
agent_template = '''
# Einzelwerkzeug-Agent
# Beginnen Sie einfach: NUR EIN Werkzeug!

import json

# Definieren Sie Ihr EINES Werkzeug
def calculator(expression: str) -> str:
    """Einen mathematischen Ausdruck auswerten."""
    try:
        # WARNUNG: eval ist gefährlich! Verwenden Sie einen ordentlichen Math-Parser in Produktion
        result = eval(expression, {"__builtins__": {}}, {})
        return str(result)
    except:
        return "Fehler: Konnte nicht berechnen"

TOOL_SCHEMA = {
    "name": "calculator",
    "description": "Einen mathematischen Ausdruck auswerten",
    "parameters": {
        "expression": "Ein mathematischer Ausdruck wie '2 + 2' oder '15 * 7'"
    }
}

def should_use_tool(user_message: str) -> bool:
    """Entscheiden, ob wir den Taschenrechner brauchen."""
    # TODO: LLM zur Klassifizierung verwenden (Kap15)
    math_keywords = ["berechnen", "rechnen", "was ist", "+", "-", "*", "/"]
    return any(kw in user_message.lower() for kw in math_keywords)

def extract_expression(user_message: str) -> str:
    """Den mathematischen Ausdruck aus der Nachricht extrahieren."""
    # TODO: LLM zum Extrahieren verwenden (Kap15)
    return user_message  # Platzhalter

def agent(user_message: str) -> str:
    """Benutzernachricht verarbeiten, möglicherweise mit Werkzeugen."""
    if should_use_tool(user_message):
        expression = extract_expression(user_message)
        result = calculator(expression)
        return f"Die Antwort ist: {result}"
    else:
        return "Ich kann bei Berechnungen helfen. Versuchen Sie, mich zu bitten, etwas zu berechnen!"

# Testen Sie es
print(agent("Was ist 15 * 7?"))
'''

print("Einzelwerkzeug-Agent Starter-Vorlage:")
print(agent_template)
```
