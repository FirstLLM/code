````python
# ===== ΠΡΟΕΤΟΙΜΑΣΙΑ =====
# Εγκατάστησε το πακέτο transformers του Hugging Face για εργασία με μεγάλα γλωσσικά μοντέλα.
!pip install -q transformers==5.23.0 torch==2.15.0  # Δοκιμασμένος συνδυασμός εκδόσεων (Παράρτημα A)

import warnings
warnings.filterwarnings('ignore')  # Μην εμφανίζεις δευτερεύουσες προειδοποιήσεις εκδόσεων.

# Φόρτωσε τα εργαλεία του Hugging Face για το GPT-2.
# Η pipeline φορτώνει το μοντέλο και αναλαμβάνει τα ενδιάμεσα βήματα.
# Σκέψου τη σαν έτοιμη ακολουθία: φόρτωση μοντέλου → επεξεργασία εισόδου → παραγωγή κειμένου.
from transformers import pipeline, logging
logging.set_verbosity_error()  # Εμφάνισε μόνο σφάλματα, όχι ενημερωτικά μηνύματα.

print('Η προετοιμασία ολοκληρώθηκε.')
````

## Πρώτη δοκιμή: παραγωγή κειμένου με το GPT-2

## Αριθμοί και μορφοποίηση
Δοκίμασε να εμφανίσεις αριθμούς μέσα σε συμβολοσειρές f. Οι τρεις γραμμές εξόδου μένουν στα αγγλικά για να τις συγκρίνεις με το παράδειγμα του βιβλίου: δείχνουν πόσες παραμέτρους έχει το GPT-2, τον ρυθμό μάθησης και το μισό μέγεθος του λεξιλογίου.

````python
# Με την κάτω παύλα διαβάζεις πιο εύκολα μεγάλους αριθμούς στην Python.
num_parameters = 124_000_000  # 124 εκατομμύρια (μέγεθος του GPT-2)
learning_rate = 0.0001        # Μικρό βήμα για την εκπαίδευση
vocab_size = 50257            # Μέγεθος του λεξιλογίου του GPT-2

# Μορφοποίηση με συμβολοσειρές f:
print(f'GPT-2 has {num_parameters:,} parameters')   # Το :, χωρίζει τις χιλιάδες με κόμμα.
print(f'Learning rate is {learning_rate:.2e}')      # Το :.2e δίνει επιστημονική γραφή.
print(f'Half the vocab: {vocab_size // 2}')         # Μέσα στα {} μπορείς να κάνεις πράξεις!
````

## Φτιάξε ένα μικρό λεξιλόγιο και έναν tokenizer
Ξεκίνα από τρεις προτάσεις και φτιάξε έναν tokenizer: ένα εργαλείο που χωρίζει το κείμενο σε μονάδες (token) και τις κωδικοποιεί με αριθμούς. Σε αυτό το παράδειγμα, κάθε λέξη είναι ένα token. Η ένδειξη `Vocabulary size` δίνει το μέγεθος του λεξιλογίου, μαζί με τα ειδικά token.

````python
# Δείγμα κειμένων που θα μπορούσες να χρησιμοποιήσεις για εκπαίδευση
texts = [
    'The secret to building AI is',
    'The key to machine learning is data',
    'AI systems learn from examples'
]

# Μάζεψε όλες τις λέξεις από όλα τα κείμενα.
all_words = []
for text in texts:
    words = text.lower().split()  # Μετέτρεψε τα γράμματα σε πεζά.
    # Η extend() προσθέτει κάθε στοιχείο ξεχωριστά στη λίστα.
    # Η append() θα πρόσθετε ολόκληρη τη λίστα ως ένα μόνο στοιχείο.
    all_words.extend(words)

print(all_words)

# Παραδείγματα επιλογής στοιχείων και τομής λίστας
print(all_words[0], all_words[-1], all_words[:3])  # Πρώτο, τελευταίο, τρία πρώτα στοιχεία

# Φτιάξε ένα λεξιλόγιο: δώσε έναν αριθμό σε κάθε διαφορετική λέξη.
vocab = {'<PAD>': 0, '<UNK>': 1}  # Δέσμευσε πρώτα τα ID των ειδικών token.

for word in all_words:
    if word not in vocab:
        # Η len(vocab) δίνει το επόμενο διαθέσιμο ID.
        # Αν το λεξικό έχει δύο εγγραφές με ID 0 και 1, το επόμενο ID είναι len(vocab) = 2.
        vocab[word] = len(vocab)

print(f'Vocabulary size: {len(vocab)}')
print(vocab)

# Αναζήτησε λέξεις στο λεξιλόγιο.
print(vocab['the'], vocab['ai'])  # Εμφάνισε τα αντίστοιχα ID.
````

## Σύγκρινε το λεξιλόγιό σου με του tokenizer του GPT-2
Δες σε τι διαφέρει ο tokenizer που χρησιμοποιεί ένα πραγματικό μοντέλο. Στις δύο γραμμές της σύγκρισης, `Our vocab` είναι το δικό μας λεξιλόγιο και `GPT-2 vocab` το λεξιλόγιο του GPT-2.

**Συμβουλές για την Python:**
- Η `dict.get(key, default)` επιστρέφει την τιμή που αντιστοιχεί στο `key` αν υπάρχει το κλειδί, αλλιώς επιστρέφει το `default`. Αντίθετα, η `dict[key]` δίνει σφάλμα αν λείπει το κλειδί.
- Με μια **περιγραφική λίστα** (*list comprehension*) φτιάχνεις μια λίστα σε μία γραμμή. Η `[expr for item in collection]` κάνει ό,τι και ένας βρόχος `for` που προσθέτει στη λίστα την τιμή της `expr` για κάθε στοιχείο.

````python
# Φόρτωσε τον tokenizer του GPT-2.
from transformers import GPT2Tokenizer
real_tok = GPT2Tokenizer.from_pretrained('gpt2')

# Σύγκρινε τα μεγέθη των λεξιλογίων.
print(f'Our vocab: {len(vocab)} words')
print(f'GPT-2 vocab: {len(real_tok)} tokens')

# Δες τι συμβαίνει με μια άγνωστη λέξη.
# Η .get(key, default) επιστρέφει το default αν δεν βρει το κλειδί, αντί να δώσει σφάλμα.
word = 'neural'
print(f"'{word}' → {vocab.get(word, vocab['<UNK>'])}")  # Δίνει το ID του <UNK>, δηλαδή 1.

# Κωδικοποίησε μια πρόταση με το λεξιλόγιό μας.
sentence = 'The neural network learns'

# Αναλυτική εκδοχή με βρόχο:
token_ids = []
for word in sentence.lower().split():
    token_id = vocab.get(word, vocab['<UNK>'])  # Πάρε το ID της λέξης ή του <UNK> αν η λέξη είναι άγνωστη.
    token_ids.append(token_id)
    print(f'  {word} → {token_id}')

print(f'ID token (βρόχος): {token_ids}')

# Σύντομη εκδοχή με περιγραφική λίστα: ίδιο αποτέλεσμα, λιγότερος κώδικας.
token_ids = [vocab.get(w, vocab['<UNK>']) for w in sentence.lower().split()]
print(f'ID token (περιγραφική λίστα): {token_ids}')
````

## Συναρτήσεις για κωδικοποίηση και αποκωδικοποίηση
Κωδικοποίησε μια πρόταση σε ID και ξαναφτιάξε το κείμενο από αυτά. Στην τελική σύγκριση, η γραμμή `Our tokens` δείχνει τα ID που έδωσε η δική σου συνάρτηση και η `GPT-2 tokens` εκείνα που έδωσε ο tokenizer του GPT-2.

````python
# Συνάρτηση: κείμενο → ID token (κωδικοποίηση)
def tokenize(text, vocab):
    words = text.lower().split()
    return [vocab.get(w, vocab['<UNK>']) for w in words]

# Συνάρτηση: ID token → κείμενο (αποκωδικοποίηση)
def detokenize(ids, vocab):
    # Φτιάξε την αντίστροφη αντιστοίχιση (ID → λέξη).
    # Περιγραφική έκφραση λεξικού: {new_key: new_val for key, val in dict.items()}
    # Η vocab.items() δίνει ζεύγη όπως ('the', 2), ('secret', 3).
    # Αντιστρέφουμε τη σειρά: (2, 'the'), (3, 'secret').
    id_to_word = {v: k for k, v in vocab.items()}
    return ' '.join(id_to_word.get(i, '<UNK>') for i in ids)

# Δοκίμασε και τις δύο κατευθύνσεις: κείμενο → ID → κείμενο.
ids = tokenize('The secret to AI', vocab)
print(f'Κωδικοποίηση: {ids}')
print(f'Αποκωδικοποίηση: {detokenize(ids, vocab)}')

# Σύγκρινε τον tokenizer μας με του GPT-2.
text = 'The secret to AI'
print(f'Our tokens:   {tokenize(text, vocab)}')
print(f'GPT-2 tokens: {real_tok.encode(text)}')  # Διαφέρουν! Το GPT-2 χρησιμοποιεί και τμήματα λέξεων, όχι μόνο ολόκληρες λέξεις.
````

## Μια απλή κλάση για τον tokenizer
Ο tokenizer που θα φτιάξεις τώρα κρατά το λεξιλόγιό του από τη μία κλήση στην επόμενη. Χειρίζεται ολόκληρες λέξεις και έχει τις μεθόδους `fit`, `encode` και `decode`.

**Οι κλάσεις της Python με λίγα λόγια:** Μια **κλάση** είναι ένα σχέδιο για τη δημιουργία αντικειμένων που συνδυάζουν δεδομένα και συναρτήσεις.

- Με το `class MyClass:` ορίζεις το σχέδιο.
- Η `__init__(self)` εκτελείται όταν δημιουργείς ένα αντικείμενο: δίνει αρχικές τιμές στα δεδομένα του.
- Το `self` αναφέρεται στο συγκεκριμένο αντικείμενο («αυτό το αυτοκίνητο», όχι «τα αυτοκίνητα γενικά»).
- Όταν καλείς μια μέθοδο μέσω ενός αντικειμένου, η Python τής περνά αυτόματα το αντικείμενο ως πρώτο όρισμα (`self`).

**Μια παρομοίωση:** Σκέψου την κλάση σαν σχέδιο αυτοκινήτου. Η `__init__` ορίζει τα αρχικά χαρακτηριστικά του, όπως το χρώμα και τον κυβισμό του κινητήρα. Όταν φτιάχνεις ένα αυτοκίνητο με βάση το σχέδιο, το `self` αναφέρεται σε εκείνο ακριβώς το αυτοκίνητο.

````python
# Αντικειμενοστρεφής tokenizer: η κλάση συνδυάζει δεδομένα και μεθόδους.
class SimpleTokenizer:
    def __init__(self):
        # Ξεκίνα το λεξιλόγιο με τα ειδικά token.
        self.word_to_id = {'<PAD>': 0, '<UNK>': 1}
        self.id_to_word = {0: '<PAD>', 1: '<UNK>'}

    def fit(self, texts):
        """Φτιάχνει το λεξιλόγιο από τα κείμενα εκπαίδευσης."""
        for text in texts:
            for word in text.lower().split():
                if word not in self.word_to_id:
                    idx = len(self.word_to_id)
                    self.word_to_id[word] = idx  # Πρόσθεσε τη νέα λέξη.
                    self.id_to_word[idx] = word  # Αντίστροφη αντιστοίχιση

    def encode(self, text):
        """Μετατρέπει το κείμενο σε ID token."""
        return [self.word_to_id.get(w, 1) for w in text.lower().split()]  # 1 = <UNK>

    def decode(self, ids):
        """Μετατρέπει τα ID token ξανά σε κείμενο."""
        return ' '.join(self.id_to_word.get(i, '<UNK>') for i in ids)

    def __len__(self):
        """Επιστρέφει το μέγεθος του λεξιλογίου ώστε να λειτουργεί η len(tok)."""
        return len(self.word_to_id)

# Δημιούργησε και εκπαίδευσε τον tokenizer.
tok = SimpleTokenizer()
tok.fit(texts)  # Φτιάξε το λεξιλόγιο από τα κείμενα εκπαίδευσης.

print(f'Vocabulary size: {len(tok)}')

# Δοκίμασε την κωδικοποίηση και την αποκωδικοποίηση.
ids = tok.encode('The secret to AI')
print(f'Κωδικοποίηση: {ids}')
print(f'Αποκωδικοποίηση: {tok.decode(ids)}')

# Κάνε άλλη μία σύγκριση με το GPT-2.
gpt2_tok = GPT2Tokenizer.from_pretrained('gpt2')
text = 'The secret to AI'
print(f'Ο tokenizer σου:  {tok.encode(text)}')
print(f'Tokenizer του GPT-2: {gpt2_tok.encode(text)}')  # Το GPT-2 χρησιμοποιεί κωδικοποίηση ζευγών byte (BPE).
````

## Επιστροφή στην αρχή: ο tokenizer σου και ο tokenizer του GPT-2
Σύγκρινε άλλη μια φορά τον tokenizer που έφτιαξες με τον tokenizer του GPT-2. Πρόσεξε πόσο μεγαλύτεροι είναι οι αριθμοί που δίνει το GPT-2 ως ID στο παράδειγμα: έχει λεξιλόγιο με 50.257 token! Αντί να βασίζεται σε ολόκληρες λέξεις, χρησιμοποιεί την κωδικοποίηση ζευγών byte (**Byte-Pair Encoding**, BPE), που επιτρέπει σε ένα token να είναι τμήμα λέξης.

````python
# Φτιάξε νέο tokenizer για να τον συγκρίνεις με του GPT-2.
my_tok = SimpleTokenizer()
my_tok.fit(['The secret to building AI is understanding'])

# Φόρτωσε τον tokenizer του GPT-2.
from transformers import GPT2Tokenizer
gpt2_tok = GPT2Tokenizer.from_pretrained('gpt2')

# Σύγκρινε τα αποτελέσματα για το ίδιο κείμενο.
text = 'The secret to AI'
print(f'Ο tokenizer σου:  {my_tok.encode(text)}')
print(f'Tokenizer του GPT-2: {gpt2_tok.encode(text)}')

print(f'\nΜέγεθος του λεξιλογίου σου:  {len(my_tok)}')
print(f'Μέγεθος του λεξιλογίου του GPT-2: {len(gpt2_tok)}')
````
